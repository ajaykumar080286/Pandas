```python
import pandas as pd
import numpy as np
```

```python
deliveries=pd.read_csv('deliveries.csv')
```

```python
deliveries.head(3)
```

```python
matches=pd.read_csv('matches - matches.csv')
```

```python
matches.head(2)
```

```python
new=deliveries.merge(matches, left_on='match_id', right_on='id')
```

```python
new.groupby(['season','batsman'])['batsman_runs'].sum().reset_index().sort_values('batsman_runs',ascending=False).reset_index().drop_duplicates(subset='season',keep='first').sort_values('season')[['season','batsman']]
```
