```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame({
    'A': ['John', 'Boby', 'Mina', 'Peter', 'Nicky'],
    'B': ['Masters', 'Graduate', 'Graduate', 'Masters', 'Graduate'],
    'C': [27, 23, 21, 23, 24]
})
```

```python
df
```

```python
table = pd.pivot_table(df, values='C', index=['A','B'], columns=['B'], aggfunc='sum')
```

```python
table
```

**Applying Pivot Table to IPL Dataset**

```python
matches=pd.read_csv('matches - matches.csv')
deliveries=pd.read_csv('deliveries.csv')
```

```python
matches
```

```python
deliveries
```

```python
mask=deliveries['batsman_runs']==6
```

```python
six=deliveries[mask]
```

```python
six.head()
```

```python
six.shape
```

```python
pt=six.pivot_table(index='over', columns='bowling_team', values='batsman_runs', aggfunc='count')
```

```python
import seaborn as sns
```

```python
sns.heatmap(pt)
```
