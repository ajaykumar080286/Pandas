```python
import pandas as pd
import numpy as np
```

```python
deliveries=pd.read_csv('deliveries.csv')
```

```python
deliveries.head(4)
```

```python
death_over=deliveries[deliveries['over']>15]
```

```python
all_bats=death_over.groupby('batsman')['batsman_runs'].count()
x=all_bats>200
batsmanlist=all_bats[x].index.tolist()
```

```python
#runs score by all these 43 batsman
#ball played by these 43 batsman
```

```python
final=deliveries[deliveries['batsman'].isin(batsmanlist)]
```

```python
runs=final.groupby('batsman')['batsman_runs'].sum()

ball=final.groupby('batsman')['batsman_runs'].count()
```

```python
sr=runs/ball*100
```
